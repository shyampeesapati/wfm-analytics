```python
import matplotlib.pyplot as plt
import numpy as np

channels = ["ATM", "Live Telephone", "Drive Through", "In Person", "Internet"]
values = [16, 2, 17, 41, 24]

# Step 1: Sort data (descending)
sorted_data = sorted(zip(values, channels), reverse=True)
values_sorted, channels_sorted = zip(*sorted_data)

# Step 2: Cumulative %
cum_values = np.cumsum(values_sorted)
cum_percent = cum_values / sum(values_sorted) * 100

# Step 3: Plot
fig, ax1 = plt.subplots()

# Bar chart
ax1.bar(channels_sorted, values_sorted)
ax1.set_ylabel("Percentage")

# Line chart (cumulative)
ax2 = ax1.twinx()
ax2.plot(channels_sorted, cum_percent, marker='o')
ax2.set_ylabel("Cumulative %")

plt.title("Pareto Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

channels = ["ATM", "Live Telephone", "Drive Through", "In Person", "Internet"]
values = [16, 2, 17, 41, 24]

# Sort ascending (ogive)
sorted_data = sorted(zip(values, channels))
values_sorted, channels_sorted = zip(*sorted_data)

cum_values = np.cumsum(values_sorted)
cum_percent = cum_values / sum(values_sorted) * 100

fig, ax1 = plt.subplots()

# Bars (make slightly transparent)
ax1.bar(channels_sorted, values_sorted, color='skyblue', alpha=0.7)
ax1.set_ylabel("Percentage")

# Line (change color + bring forward)
ax2 = ax1.twinx()
ax2.plot(channels_sorted, cum_percent, color='red', marker='o', linewidth=2, zorder=5)
ax2.set_ylabel("Cumulative %")

plt.title("Bar Chart with Ogive")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

channels = ["ATM", "Live Telephone", "Drive Through", "In Person", "Internet"]
values = [16, 2, 17, 41, 24]

# Sort descending (Pareto)
sorted_data = sorted(zip(values, channels), reverse=True)
values_sorted, channels_sorted = zip(*sorted_data)

# Cumulative %
cum_values = np.cumsum(values_sorted)
cum_percent = cum_values / sum(values_sorted) * 100

fig, ax1 = plt.subplots()

# Bars (lighter)
ax1.bar(channels_sorted, values_sorted, color='skyblue', alpha=0.7)
ax1.set_ylabel("Percentage")

# Cumulative line (fixed visibility)
ax2 = ax1.twinx()
ax2.plot(channels_sorted, cum_percent,
         color='orange', marker='o', linewidth=2, markersize=8, zorder=5)
ax2.set_ylabel("Cumulative %")

plt.title("Pareto Chart")
plt.show()
```

```python
day = [16,17,17,18,18,18,19,19,20,20,21,22,22,25,27,32,38,42]
night = [18,18,19,19,20,21,23,28,32,33,41,45]

def stem_leaf(data, title):
    print(f"\n{title}")
    stems = {}
    
    for num in sorted(data):
        stem = num // 10
        leaf = num % 10
        stems.setdefault(stem, []).append(leaf)
    
    for stem, leaves in stems.items():
        print(f"{stem} | {' '.join(map(str, leaves))}")

# Call for both
stem_leaf(day, "Day Students")
stem_leaf(night, "Night Students")
```
