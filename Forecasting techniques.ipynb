# Forecasting techniques- #
**Naive forecasting:**

Naive forecasting is the simplest of all forecasting methods, it is taking the previous actual value as forecast for the next period

$$α=\frac{1}{n−1}​∑^n_{t=2}​(Y_{t}​−Y_{t−1}​)$$

So lets say, we have 5 data points 110, 120, 130, 140 & 150, then take 4 differences, because t cannot be 1, then you get 4 10s, now n is 5, minus 1 is 4, 40/4 is 10

The above method is **Random walk with Drift**, In a **seasonal random walk**, For Jan forecast, you wouldnt consider previous year Dec forecast but previous Jan forecast.  So far, we have covered Random walk or Naive forecast, Seasonal random walk and Random walk with drift.

**Moving Average Method:**

Denoted as MA(n).  This average is the mean of defined number of previous data points in succession, when we move to next data point, the moving average moves as well

**Weighted Moving Average method:**  Denoted as WMA(n).  In this method, weights are given to past values instead of assuming equal weightage.  The total of weights is either 1 or 100%

While we havent discussed **Mean or Simple average**, it is the basic forecasting technique.
