Welcome to the world of business statistics!  
Statistics is the branch of mathematics that transforms data into useful information for decision makers. 
***
Introduction to Statistics and Data  
Statistics is of two types, they are Descriptive Statistics & Inferential Statistics.  
•Descriptive Statistics is collecting, summarizing and describing data.  
•Inferential Statistics is drawing conclusions and/or making decisions concerning a population based only on Sample data.  
Inferential Statistics is of two types. They are Estimation and Hypothesis Testing.  
•Estimation: Calculating Population parameter from Sample statistic  
•Hypothesis Testing: Agreeing on an assumption and then checking if the assumption is valid or not using sample data  
***  
**Key definitions:**  
•Variable: Characteristic of an element or individual or organization or entity  
•Population: All items about which you want to draw a conclusion  
•Sample: Portion of population selected for analysis. It has to be a representative  
•Parameter: Numerical measure of the characteristic of a population  
•Statistic: Numerical measure that describes the characteristic of a sample  
***
**Benefits of Sample-** Saves time and cost, population is too large in some cases and when we need to consider replacement like product testing.  
***
**Sources of data:**  
Primary Sources: Data collector is the one using the data for analysis  
•Data from a political survey  
•Data collected for an experiment  
•Observed data  
Secondary Sources: Person performing the analysis is not the data collector  
•Analyzing census data  
•Analyzing data from print journals or data already published in internet  
***
**Types of Variables:**  
•Categorical, qualitative variables that can be placed into categories like true or false, non metric  
•Numerical, quantitative variables represent quantities, metric  
Numerical variables are of 2 types, they are:  
•Discrete, can be counted  
•Continous data, can be measured  
***
**Scales of Measurement**
•Nominal, Categorical, labels or names used to identify an attribute, numeric code can be nominal  
•Ordinal, Categorical, difference is not measured, same as nominal plus rank or order is indicated  
•Interval scale, metric, uses fixed unit of measure, has the properties of ordinal scale, true 0 does not exist(temperature, dates)  
•Ratio scale, metric, expressed as ratio of values from interval scale, True 0 exists (absence of quantity)(height, weight, income)  
Note: Temperature and dates are interval scale as differences can be measured but the 0 point is not defined like height or weight measured in Cms or Kgs where at 0 point, there is no defined value.  
***
**Types of data:**  
•Qualitative data: Categorical data, Nominal and Ordinal scales generate this data  
•Quantitative data: Numerical data, Interval or Ratio scales generate this data  
•Cross sectional data: Data collected at a point in time, Buildings approved in 2025 across the cities  
•Time series data: Data collected over a period of time, buildings approved in Hyderabad from 2000 to 2025
